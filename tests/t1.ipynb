```python
import torch

x = torch.tensor(2.0, requires_grad=True)
y = x ** 2

print(y.requires_grad)  # True

y.backward()
print(x.grad)          # tensor(4.)
```

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

with torch.no_grad():
    y = x**2

print(y.requires_grad)  # False
```
